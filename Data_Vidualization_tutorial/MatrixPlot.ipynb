```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## Heat Map

```python
flights = sns.load_dataset('flights')
flights
```

```python
tip = sns.load_dataset('tips')
tip
```

```python
tipscorr = tip[['total_bill','tip','size']]
```

```python
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True)
```

### Cluster map

```python
sns.clustermap(tipscorr.corr(),figsize=(6,4), annot=True)
```

## pivot table with head map

```python
pvflight = flights.pivot_table(values='passengers', index='month', columns='year')
```

```python
pvflight
```

```python
sns.heatmap(pvflight, cmap='rainbow')
plt.title('Flights Data')
```
