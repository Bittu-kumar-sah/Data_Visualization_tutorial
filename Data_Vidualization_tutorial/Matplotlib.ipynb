```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

### Matplotlib Basics

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 2 # Broadcasting
y
```

```python
plt.plot(x,y)
plt.title('X-Y Comparision')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
```

### Subplots in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y**2,y**2)
plt.subplot(2,2,4)
plt.plot(y,x)
```

### Object Oriented Way

```python
fig = plt.figure(figsize=(20,8), dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.4,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```

### Types of Plots

```python
plt.scatter(x,y)
```

```python
from random import sample
data = sample(range(1,1000),100)
plt.hist(data)
```

```python
plt.boxplot(x)
```

### Saving Plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig('BasicPlot.jpg')
```
